```python
import pandas as pd

df = pd.read_csv("customer_shopping_data.csv")
df.head()
```

```python
print(df.columns)
```

```python
import matplotlib.pyplot as plt

category_sales = df.groupby('category')['price'].sum()

category_sales.plot(kind='bar')
plt.title('Sales by Category')
plt.xlabel('Category')
plt.ylabel('Total Sales')
plt.show()
```

```python
payment_sales = df.groupby('payment_method')['price'].sum()

payment_sales.plot(kind='pie', autopct='%1.1f%%')
plt.title('Sales by Payment Method')
plt.ylabel('')
plt.show()
```

```python
df['gender'].value_counts().plot(kind='bar')
plt.title('Gender Distribution')
plt.xlabel('Gender')
plt.ylabel('Count')
plt.show()
```

```python
top_categories = df.groupby('category')['price'].sum().sort_values(ascending=False)

top_categories.plot(kind='bar')
plt.title('Top Categories by Sales')
plt.xlabel('Category')
plt.ylabel('Sales')
plt.show()
```
